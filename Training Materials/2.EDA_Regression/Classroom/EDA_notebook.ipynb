## Correlation b/w Variables

```python
#iris_df.corr()
iris_df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(15,9))
sns.heatmap(iris_df.corr(numeric_only=True), cmap='Greens', annot = True);
plt.show()
```

```python
# Inspect correlations: show top positive/negative correlations (flattened)
corr = iris_df.corr(numeric_only=True).abs()
corr_unstack = corr.unstack().sort_values(ascending=False)
# filter out self correlations
corr_pairs = corr_unstack[corr_unstack < 1].drop_duplicates()
print('Top correlated variable pairs (absolute correlation):')
print(corr_pairs.head(10))
```

## Variation in Bandwidth

```python

# Large bandwidth
sns.kdeplot(iris_df['sepalwidth'], shade=True, bw=0.5, color="olive")
plt.show()
```

```python
# Smaller bandwidth
sns.kdeplot(iris_df['sepalwidth'], shade=True, bw=0.05, color="olive")
plt.show()
```

```python
# KDE plots for petal length and petal width
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.kdeplot(iris_df['petallength'], shade=True)
plt.title('KDE - Petal Length')
plt.subplot(1,2,2)
sns.kdeplot(iris_df['petalwidth'], shade=True)
plt.title('KDE - Petal Width')
plt.show()
```

## Box Plot

```python
sns.boxplot(x='petalwidth', data=iris_df)
plt.title('Boxplot for Petal Width')
plt.show()
```

```python
# Box plots for other numeric variables
fig, axes = plt.subplots(1,3, figsize=(18,5))
sns.boxplot(x='sepallength', data=iris_df, ax=axes[0])
axes[0].set_title('Sepal Length')
sns.boxplot(x='sepalwidth', data=iris_df, ax=axes[1])
axes[1].set_title('Sepal Width')
sns.boxplot(x='petallength', data=iris_df, ax=axes[2])
axes[2].set_title('Petal Length')
plt.show()

# Simple IQR-based outlier counts
for col in ['sepallength','sepalwidth','petallength','petalwidth']:
    q1 = iris_df[col].quantile(0.25)
    q3 = iris_df[col].quantile(0.75)
    iqr = q3 - q1
    outliers = iris_df[(iris_df[col] < (q1 - 1.5*iqr)) | (iris_df[col] > (q3 + 1.5*iqr))]
    print(f'{col}: {len(outliers)} outlier(s)')
```

```python
# Per-class summary statistics
display(iris_df.groupby('class').mean().transpose())
```

```python
# Boxplots per class (alternate layout)
fig, axes = plt.subplots(2,2, figsize=(14,10))
sns.boxplot(y='sepallength', x='class', data=iris_df, ax=axes[0,0])
sns.boxplot(y='sepalwidth', x='class', data=iris_df, ax=axes[0,1])
sns.boxplot(y='petallength', x='class', data=iris_df, ax=axes[1,0])
sns.boxplot(y='petalwidth', x='class', data=iris_df, ax=axes[1,1])
plt.tight_layout()
plt.show()
```

```python
# Encode class to numeric and show correlation with features (useful before modeling)
iris_encoded = iris_df.copy()
iris_encoded['class_encoded'] = iris_encoded['class'].astype('category').cat.codes
display(iris_encoded.corr(numeric_only=True)['class_encoded'].sort_values(ascending=False))
```

## Contour Plot

```python
import matplotlib.pyplot as plt
import numpy as np
  
feature_x = np.arange(0, 50, 2)
feature_y = np.arange(0, 50, 3)
  
# Creating 2-D grid of features
[X, Y] = np.meshgrid(feature_x, feature_y)
  
fig, ax = plt.subplots(1, 1)
  
Z = np.cos(X / 2) + np.sin(Y / 4)
  
# plots contour lines
ax.contour(X, Y, Z)
  
ax.set_title('Contour Plot')
ax.set_xlabel('feature_x')
ax.set_ylabel('feature_y')
  
plt.show()
```
