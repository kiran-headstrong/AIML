```python
# python
import numpy as np
import pandas as pd
from math import ceil

data=df
# set a reproducible random seed
RSEED = 42
rng = np.random.default_rng(RSEED)

# Helper to display selected samples
def show_sample(df, name):
    print(f"\n--- {name} (n={len(df)}) ---")
    print(df[['patient', 'bp_before']].to_string(index=False))

# 1) Simple random sampling (without replacement)
def simple_random_sample(df, n, random_state=RSEED):
    return df.sample(n=n, replace=False, random_state=random_state)

srs_n = 3
srs_sample = simple_random_sample(data, srs_n)
show_sample(srs_sample, "Simple Random Sample")

# 2) Stratified sampling
# If you have a natural strata column (e.g., 'agegrp' or 'sex'), use it.
# In this dataset 'agegrp' and 'sex' are constant, so we demonstrate creating strata by binning bp_before.
def stratified_sample(df, strata_col, n_per_stratum=None, frac_per_stratum=None, random_state=RSEED):
    rng = np.random.RandomState(random_state)
    grouped = df.groupby(strata_col)
    samples = []
    if n_per_stratum is not None:
        for name, group in grouped:
            k = min(n_per_stratum, len(group))
            samples.append(group.sample(n=k, random_state=rng))
    elif frac_per_stratum is not None:
        for name, group in grouped:
            k = max(1, int(round(len(group) * frac_per_stratum)))
            samples.append(group.sample(n=k, random_state=rng))
    else:
        raise ValueError("Either n_per_stratum or frac_per_stratum must be provided.")
    return pd.concat(samples).reset_index(drop=True)

# Create strata by binning bp_before into 2 strata (low/high)
data['bp_strata'] = pd.qcut(data['bp_before'], q=2, labels=['low','high'])
# request 1 sample per strata
strat_sample = stratified_sample(data, 'bp_strata', n_per_stratum=1)
show_sample(strat_sample, "Stratified Sample (1 per bp_strata)")

# 3) Cluster sampling
# If you have a cluster column (e.g., clinic id), use it. Here we demonstrate by creating cluster IDs.
# Create 3 artificial clusters by patient id modulo 3
data['cluster_id'] = data['patient'] % 3  # cluster labels 0,1,2
def cluster_sample(df, cluster_col, n_clusters_to_sample, random_state=RSEED, sample_within_cluster=None):
    rng = np.random.RandomState(random_state)
    clusters = df[cluster_col].unique()
    chosen_clusters = rng.choice(clusters, size=min(n_clusters_to_sample, len(clusters)), replace=False)
    subset = df[df[cluster_col].isin(chosen_clusters)]
    if sample_within_cluster is None:
        # take all units in chosen clusters (one-stage cluster sample)
        return subset.reset_index(drop=True)
    else:
        # two-stage: sample a fixed number per chosen cluster (or fraction)
        result = []
        for cl in chosen_clusters:
            grp = df[df[cluster_col] == cl]
            k = sample_within_cluster
            k = min(k, len(grp))
            result.append(grp.sample(n=k, random_state=rng))
        return pd.concat(result).reset_index(drop=True)

cluster_sample_1 = cluster_sample(data, 'cluster_id', n_clusters_to_sample=2, random_state=RSEED)
show_sample(cluster_sample_1, "Cluster Sample (one-stage, select 2 clusters)")

# Two-stage cluster: select 1 cluster, then sample up to 2 units within it
cluster_sample_2 = cluster_sample(data, 'cluster_id', n_clusters_to_sample=1, random_state=RSEED, sample_within_cluster=2)
show_sample(cluster_sample_2, "Cluster Sample (two-stage: 1 cluster, 2 units within)")

# 4) Systematic sampling
# Choose sample size n_sys and select every k-th after a random start in 0..k-1
def systematic_sample(df, n, random_state=RSEED):
    N = len(df)
    if n <= 0 or n > N:
        raise ValueError("n must be between 1 and N")
    # compute step as floor(N/n)
    step = int(np.floor(N / n))
    if step < 1:
        step = 1
    rng = np.random.RandomState(random_state)
    # use randint for RandomState
    start = rng.randint(0, step)  # random start between 0 and step-1 inclusive
    indices = list(range(start, N, step))
    # If we got fewer than n due to floor rounding, extend cyclically
    i = 0
    while len(indices) < n:
        candidate = (indices[-1] + 1 + i) % N
        if candidate not in indices:
            indices.append(candidate)
        i += 1
    indices = indices[:n]
    return df.iloc[indices].reset_index(drop=True)

sys_n = 3
sys_sample = systematic_sample(data, sys_n)
show_sample(sys_sample, "Systematic Sample (n=3)")

# --- Summary: show samples side-by-side if desired ---
print("\nSummary counts:")
print(f"Original N = {len(data)}")
print(f"SRS n = {len(srs_sample)}")
print(f"Stratified n = {len(strat_sample)}")
print(f"Cluster (one-stage) n = {len(cluster_sample_1)}")
print(f"Cluster (two-stage) n = {len(cluster_sample_2)}")
print(f"Systematic n = {len(sys_sample)}")
```

```python
# python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis, probplot

sns.set(style="whitegrid")

# Prepare data (drop NaNs and ensure numeric)
before = pd.to_numeric(df['bp_before'], errors='coerce').dropna().values
after = pd.to_numeric(df['bp_after'], errors='coerce').dropna().values

def stats_summary(x):
    return {
        'n': len(x),
        'mean': np.mean(x),
        'median': np.median(x),
        'std': np.std(x, ddof=1),
        'variance': np.var(x, ddof=1),
        'skew': skew(x, bias=False),
        'kurtosis_excess': kurtosis(x, fisher=True, bias=False),
        'iqr': np.percentile(x, 75) - np.percentile(x, 25),
        'mad': np.mean(np.abs(x - np.mean(x))),
        'range': np.ptp(x)
    }

s_before = stats_summary(before)
s_after = stats_summary(after)

# Print summaries
print("BP Before summary:")
for k, v in s_before.items():
    print(f"  {k}: {v}")
print("\nBP After summary:")
for k, v in s_after.items():
    print(f"  {k}: {v}")

# Function to compute bootstrap kurtosis estimates (gives uncertainty)
def bootstrap_kurtosis(x, n_boot=1000, random_state=0):
    rng = np.random.RandomState(random_state)
    boots = []
    n = len(x)
    for _ in range(n_boot):
        sample = rng.choice(x, size=n, replace=True)
        boots.append(kurtosis(sample, fisher=True, bias=False))
    return np.array(boots)

# Compute bootstrap kurtosis for each
boots_before = bootstrap_kurtosis(before, n_boot=500)
boots_after = bootstrap_kurtosis(after, n_boot=500)

# Plotting grid: 2 rows (before, after) x 4 columns (hist+kde, box+violin, Q-Q, kurtosis)
fig, axes = plt.subplots(2, 4, figsize=(18, 8))

# Helper to annotate histogram
def annotate_stats(ax, stats):
    textstr = (f"n={stats['n']}\nmean={stats['mean']:.2f}\nstd={stats['std']:.2f}\n"
               f"skew={stats['skew']:.3f}\nkurtosis={stats['kurtosis_excess']:.3f}")
    ax.text(0.98, 0.95, textstr, transform=ax.transAxes, fontsize=9,
            verticalalignment='top', horizontalalignment='right',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

# Row 0: before
ax = axes[0, 0]
sns.histplot(before, kde=True, bins=25, color="C0", ax=ax)
ax.set_title("BP Before: Histogram + KDE")
ax.axvline(s_before['mean'], color="C1", linestyle="--", label=f"mean={s_before['mean']:.2f}")
ax.axvline(s_before['median'], color="C2", linestyle=":", label=f"median={s_before['median']:.2f}")
ax.legend()
annotate_stats(ax, s_before)

ax = axes[0, 1]
sns.violinplot(data=before, inner="quartile", color="C3", ax=ax, orient="h")
sns.boxplot(data=before, color="C4", ax=ax, orient="h", width=0.15)
ax.set_title("BP Before: Violin + Boxplot")

ax = axes[0, 2]
probplot(before, dist="norm", plot=axes[0, 2])
axes[0, 2].set_title("BP Before: Q-Q plot")

# Kurtosis subplot for 'before' (bar + bootstrap histogram inset)
ax = axes[0, 3]
# main bar showing point estimate
ax.bar(0, s_before['kurtosis_excess'], color="C5", width=0.6)
ax.set_xlim(-0.5, 1.5)
ax.set_xticks([0])
ax.set_xticklabels([f"kurtosis\n(before)"])
ax.set_ylabel("excess kurtosis")
ax.axhline(0, color="k", linestyle="--", linewidth=0.8)
# annotate value
ax.text(0, s_before['kurtosis_excess'], f"{s_before['kurtosis_excess']:.3f}",
        va="bottom", ha="center", fontsize=10, fontweight="bold")
# inset: histogram of bootstrap kurtosis to show uncertainty
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
axins = inset_axes(ax, width="60%", height="35%", loc='upper right', borderpad=1)
axins.hist(boots_before, bins=25, color="C5", alpha=0.7)
axins.axvline(np.mean(boots_before), color="k", linestyle="--", linewidth=0.8)
axins.set_xticks([])
axins.set_yticks([])
axins.set_title("bootstrap", fontsize=8)

# Row 1: after
ax = axes[1, 0]
sns.histplot(after, kde=True, bins=25, color="C0", ax=ax)
ax.set_title("BP After: Histogram + KDE")
ax.axvline(s_after['mean'], color="C1", linestyle="--", label=f"mean={s_after['mean']:.2f}")
ax.axvline(s_after['median'], color="C2", linestyle=":", label=f"median={s_after['median']:.2f}")
ax.legend()
annotate_stats(ax, s_after)

ax = axes[1, 1]
sns.violinplot(data=after, inner="quartile", color="C3", ax=ax, orient="h")
sns.boxplot(data=after, color="C4", ax=ax, orient="h", width=0.15)
ax.set_title("BP After: Violin + Boxplot")

ax = axes[1, 2]
probplot(after, dist="norm", plot=axes[1, 2])
axes[1, 2].set_title("BP After: Q-Q plot")

# Kurtosis subplot for 'after'
ax = axes[1, 3]
ax.bar(0, s_after['kurtosis_excess'], color="C6", width=0.6)
ax.set_xlim(-0.5, 1.5)
ax.set_xticks([0])
ax.set_xticklabels([f"kurtosis\n(after)"])
ax.set_ylabel("excess kurtosis")
ax.axhline(0, color="k", linestyle="--", linewidth=0.8)
ax.text(0, s_after['kurtosis_excess'], f"{s_after['kurtosis_excess']:.3f}",
        va="bottom", ha="center", fontsize=10, fontweight="bold")
axins = inset_axes(ax, width="60%", height="35%", loc='upper right', borderpad=1)
axins.hist(boots_after, bins=25, color="C6", alpha=0.7)
axins.axvline(np.mean(boots_after), color="k", linestyle="--", linewidth=0.8)
axins.set_xticks([])
axins.set_yticks([])
axins.set_title("bootstrap", fontsize=8)

plt.tight_layout()
plt.show()

# Optional: Overlaid distributions
plt.figure(figsize=(10, 4))
sns.histplot(before, color="C0", label="before", kde=True, stat="density", bins=25, alpha=0.4)
sns.histplot(after, color="C1", label="after", kde=True, stat="density", bins=25, alpha=0.4)
plt.title("Overlaid distributions: BP Before vs After")
plt.legend()
plt.show()

# Paired differences if arrays are same length
if len(before) == len(after):
    diff = after - before
    plt.figure(figsize=(10, 3.5))
    sns.histplot(diff, kde=True, color="C2", bins=25)
    plt.axvline(np.mean(diff), color="C1", linestyle="--", label=f"mean diff={np.mean(diff):.2f}")
    plt.title("Distribution of (after - before) [paired]")
    plt.legend()
    plt.show()
```

```python
# python
import pandas as pd

# Create sample data
data_prob = {
    'visited_product_page': [True, False, True, True, False, False, True, True, False, True],
    'made_purchase':        [True, False, False, True, False, True, False, True, False, False]
}
df_prob = pd.DataFrame(data_prob)

# Show the data
print("Sample Visitor Data:")
print(df_prob.to_string(index=False))
print()

# Use df_prob (DataFrame) and booleans directly
n_total = len(df_prob)

# Counts using boolean sums
n_visited_page = df_prob['visited_product_page'].sum()        # counts True
n_purchased = df_prob['made_purchase'].sum()
n_visited_and_purchased = ((df_prob['visited_product_page']) & (df_prob['made_purchase'])).sum()

# Probabilities
P_visited = n_visited_page / n_total
P_purchased = n_purchased / n_total
P_visited_and_purchased = n_visited_and_purchased / n_total

# Conditional probabilities
P_purchased_given_visited = (n_visited_and_purchased / n_visited_page) if n_visited_page > 0 else float('nan')
P_visited_given_purchased = (n_visited_and_purchased / n_purchased) if n_purchased > 0 else float('nan')

# Check independence: P(A and B) == P(A)*P(B)
independent = abs(P_visited_and_purchased - (P_visited * P_purchased)) < 1e-12

# Print nicely
print(f"Total observations: {n_total}")
print(f"P(Visited Page) = {P_visited:.3f}   (count={n_visited_page})")
print(f"P(Purchased) = {P_purchased:.3f}   (count={n_purchased})")
print(f"P(Visited and Purchased) = {P_visited_and_purchased:.3f}   (count={n_visited_and_purchased})")
print()
print(f"P(Purchased | Visited) = {P_purchased_given_visited:.3f}")
print(f"P(Visited | Purchased) = {P_visited_given_purchased:.3f}")
print()
print("Independence check:")
if independent:
    print("  Event 'Visited' and 'Purchased' appear independent (P(A∩B) ≈ P(A)P(B)).")
else:
    print("  Events are NOT independent (P(A∩B) != P(A)P(B)).")
```

```python
#3.2Bayes Theorem
#https://snowch.github.io/learn_probability/chapter_05.html
```

```python
# Parameters
p_disease = 0.01        # P(D) - Prior probability (prevalence)
p_pos_given_disease = 0.95 # P(Pos|D) - Sensitivity
p_neg_given_no_disease = 0.95 # P(Neg|D^c) - Specificity

# Derived probabilities
p_no_disease = 1 - p_disease                 # P(D^c)
p_pos_given_no_disease = 1 - p_neg_given_no_disease # P(Pos|D^c) - False Positive Rate

# Calculate P(Pos) using Law of Total Probability
p_pos = (p_pos_given_disease * p_disease) + (p_pos_given_no_disease * p_no_disease)

# Calculate P(D|Pos) using Bayes' Theorem
p_disease_given_pos = (p_pos_given_disease * p_disease) / p_pos

print(f"Prior P(Disease): {p_disease:.4f}")
print(f"Sensitivity P(Pos|Disease): {p_pos_given_disease:.4f}")
print(f"Specificity P(Neg|No Disease): {p_neg_given_no_disease:.4f}")
print(f"False Positive Rate P(Pos|No Disease): {p_pos_given_no_disease:.4f}")
print("-" * 30)
print(f"Overall P(Pos): {p_pos:.4f}")
print(f"Posterior P(Disease|Pos): {p_disease_given_pos:.4f}")

# What if the test is *negative*? Calculate P(Disease | Neg)
# P(Neg) = P(Neg|D)P(D) + P(Neg|D^c)P(D^c)
p_neg_given_disease = 1 - p_pos_given_disease # P(Neg|D) - False Negative Rate
p_neg = (p_neg_given_disease * p_disease) + (p_neg_given_no_disease * p_no_disease)

# P(D|Neg) = P(Neg|D)P(D) / P(Neg)
p_disease_given_neg = (p_neg_given_disease * p_disease) / p_neg

print("-" * 30)
print(f"Overall P(Neg): {p_neg:.4f}")
print(f"Posterior P(Disease|Neg): {p_disease_given_neg:.4f}")
print(f"Posterior P(No Disease|Neg) = {1 - p_disease_given_neg:.4f}")
```

```python
# 2.1 Hypothesis testing  
import numpy as np
import pandas as pd
from scipy import stats
from scipy.stats import shapiro, ttest_rel, wilcoxon
import matplotlib.pyplot as plt

# Prepare paired data: ensure numeric and aligned (drop rows with NA in either)
paired = df[['bp_before', 'bp_after']].apply(pd.to_numeric, errors='coerce').dropna()
before = paired['bp_before'].values
after = paired['bp_after'].values
n = len(paired)
print(f"Number of paired observations: {n}")

# Differences
diff = after - before
mean_diff = np.mean(diff)
median_diff = np.median(diff)
std_diff = np.std(diff, ddof=1)

print(f"\nMean difference (after - before): {mean_diff:.4f}")
print(f"Median difference: {median_diff:.4f}")
print(f"Std of differences: {std_diff:.4f}")

# 1) Check normality: Shapiro-Wilk on before, after, and diff
def shapiro_test(x, label):
    stat, p = shapiro(x)
    print(f"Shapiro-Wilk for {label}: W={stat:.4f}, p={p:.4f} "
          + ("(reject normality at alpha=0.05)" if p < 0.05 else "(do not reject normality)"))

shapiro_test(before, 'bp_before')
shapiro_test(after, 'bp_after')
shapiro_test(diff, 'difference (after-before)')

# 2) Choose test: paired t-test if diff is approx normal, else Wilcoxon signed-rank
alpha = 0.05
# Heuristic: use Shapiro p-value for diff
_, p_shapiro_diff = shapiro(diff)
if p_shapiro_diff > 0.05 and n >= 10:
    # paired t-test
    tstat, pval = ttest_rel(after, before)
    print("\nPaired t-test result:")
    print(f"  t = {tstat:.4f}, p = {pval:.6f}")
    if pval < alpha:
        print("  Reject H0: difference is statistically significant (two-sided).")
    else:
        print("  Fail to reject H0: no evidence of difference (two-sided).")
else:
    # Wilcoxon signed-rank test (nonparametric)
    # Use zero_method='wilcox' (excludes zero differences) default in newer scipy
    try:
        wstat, pval = wilcoxon(after, before)
        print("\nWilcoxon signed-rank test result:")
        print(f"  W = {wstat:.4f}, p = {pval:.6f}")
        if pval < alpha:
            print("  Reject H0: difference is statistically significant (two-sided).")
        else:
            print("  Fail to reject H0: no evidence of difference (two-sided).")
    except ValueError as e:
        print("\nWilcoxon test could not be performed:", e)
        print("  (This can happen if all differences are zero or sample too small.)")

# 3) Effect size: paired Cohen's d (mean diff / sd of differences)
cohens_d = mean_diff / std_diff if std_diff != 0 else np.nan
print(f"\nPaired Cohen's d = {cohens_d:.4f} (small~0.2, medium~0.5, large~0.8)")

# 4) Bootstrap 95% CI for mean difference
def bootstrap_ci_mean_diff(x, n_boot=5000, ci=95, random_state=0):
    rng = np.random.RandomState(random_state)
    n = len(x)
    boots = np.empty(n_boot)
    for i in range(n_boot):
        sample = rng.choice(x, size=n, replace=True)
        boots[i] = np.mean(sample)
    lower = np.percentile(boots, (100-ci)/2)
    upper = np.percentile(boots, 100-(100-ci)/2)
    return lower, upper, boots

lower, upper, boots = bootstrap_ci_mean_diff(diff, n_boot=5000)
print(f"\nBootstrap {95}% CI for mean difference: [{lower:.4f}, {upper:.4f}]")

# 5) Quick plots to accompany the tests
import seaborn as sns
sns.set(style="whitegrid")

plt.figure(figsize=(10,3))
plt.subplot(1,3,1)
sns.histplot(diff, kde=True, bins=25, color="C2")
plt.title("Histogram of (after - before)")
plt.axvline(mean_diff, color="k", linestyle="--", label=f"mean={mean_diff:.2f}")
plt.legend()

plt.subplot(1,3,2)
sns.boxplot(x=diff, color="C3")
plt.title("Boxplot of differences")

plt.subplot(1,3,3)
# Bootstrap distribution of mean
sns.histplot(boots, bins=30, color="C4", kde=True)
plt.axvline(lower, color="k", linestyle="--", label=f"CI lower={lower:.2f}")
plt.axvline(upper, color="k", linestyle="--", label=f"CI upper={upper:.2f}")
plt.title("Bootstrap distribution of mean difference")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
#2.2Regression and Correlation Analysis
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm
import statsmodels.formula.api as smf

sns.set(style="whitegrid")

# 1) Load the data (you can also read from CSV; here we define the data directly)
data = df


# 2) Correlation analysis
x = data['bp_before']
y = data['bp_after']

pearson_r, pearson_p = stats.pearsonr(x, y)
spearman_rho, spearman_p = stats.spearmanr(x, y)

print("\nCorrelation results:")
print(f" Pearson r = {pearson_r:.4f}, p = {pearson_p:.4f}")
print(f" Spearman rho = {spearman_rho:.4f}, p = {spearman_p:.4f}")

# 3) Simple linear regression: bp_after ~ bp_before using OLS
X = sm.add_constant(x)  # adds intercept term
model = sm.OLS(y, X).fit()
print("\nOLS regression summary:")
print(model.summary())

# Extract key regression outputs
intercept, slope = model.params['const'], model.params['bp_before']
r2 = model.rsquared
print(f"\nRegression equation: bp_after = {intercept:.3f} + {slope:.3f} * bp_before")
print(f"R-squared = {r2:.3f}")

# 4) Confidence intervals and p-values for coefficients
conf_int = model.conf_int().loc[['const','bp_before']]
pvals = model.pvalues
stderr = model.bse
print("\nCoefficients, SE, p-values, 95% CI:")
for coef in ['const','bp_before']:
    print(f" {coef}: coef={model.params[coef]:.4f}, SE={stderr[coef]:.4f}, p={pvals[coef]:.4f}, CI=({conf_int.loc[coef,0]:.4f}, {conf_int.loc[coef,1]:.4f})")

# 5) Plot scatter with regression line and annotate Pearson r
plt.figure(figsize=(6,5))
sns.regplot(x='bp_before', y='bp_after', data=data, ci=95, scatter_kws={'s':60})
plt.title(f"bp_after vs bp_before (Pearson r={pearson_r:.3f}, p={pearson_p:.3f})")
plt.xlabel("bp_before")
plt.ylabel("bp_after")
plt.tight_layout()
plt.show()

# 6) Residual diagnostics
fitted = model.fittedvalues
resid = model.resid

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.scatter(fitted, resid, edgecolor='k')
plt.axhline(0, color='gray', linestyle='--')
plt.xlabel("Fitted values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted")


# optionally add a lowess smooth
from statsmodels.nonparametric.smoothers_lowess import lowess
smoothed = lowess(resid, fitted)
plt.plot(smoothed[:,0], smoothed[:,1], color='red')

plt.subplot(1,2,2)
sm.qqplot(resid, line='45', fit=True, ax=plt.gca())
plt.title("Q-Q plot of residuals")
plt.tight_layout()
plt.show()

# 7) Robust regression (Huber) as an alternative
robust_model = smf.rlm("bp_after ~ bp_before", data=data, M=sm.robust.norms.HuberT()).fit()
print("\nRobust regression (Huber) summary:")
print(robust_model.summary())

# Plotting the regression line
plt.scatter(x, y, color='blue')
plt.plot(x, model.predict(X), color='red')
plt.title('Regression Analysis')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
# 2.3 Confidennce Interval
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from scipy import stats

sns.set(style="whitegrid")


# Prepare variables
x = data['bp_before'].astype(float)
y = data['bp_after'].astype(float)
n = len(data)
alpha = 0.05

# Fit OLS regression
X = sm.add_constant(x)
model = sm.OLS(y, X).fit()
params = model.params
se = model.bse
df_reg = n - 2

# Critical values
z_crit = stats.norm.ppf(1 - alpha/2)
t_crit_reg = stats.t.ppf(1 - alpha/2, df=df_reg)

# Compute Normal and t CIs for regression coefficients
reg_ci = {}
for name in ['const', 'bp_before']:
    est = params[name]
    se_ = se[name]
    ci_z = (est - z_crit * se_, est + z_crit * se_)
    ci_t = (est - t_crit_reg * se_, est + t_crit_reg * se_)
    reg_ci[name] = {'est': est, 'se': se_, 'ci_z': ci_z, 'ci_t': ci_t}

# Paired differences: mean, se, CIs
diff = (data['bp_after'] - data['bp_before']).values
mean_diff = diff.mean()
sd_diff = diff.std(ddof=1)
se_mean_diff = sd_diff / np.sqrt(n)
df_diff = n - 1
t_crit_diff = stats.t.ppf(1 - alpha/2, df=df_diff)

ci_z_diff = (mean_diff - z_crit * se_mean_diff, mean_diff + z_crit * se_mean_diff)
ci_t_diff = (mean_diff - t_crit_diff * se_mean_diff, mean_diff + t_crit_diff * se_mean_diff)

# Print regression results and intervals
print("Regression coefficients and 95% CIs (Normal z vs Student-t):")
for name, info in reg_ci.items():
    print(f" {name}: est={info['est']:.4f}, SE={info['se']:.4f}")
    print(f"    Normal z CI (95%): [{info['ci_z'][0]:.4f}, {info['ci_z'][1]:.4f}] (z={z_crit:.4f})")
    print(f"    t CI    (95%, df={df_reg}): [{info['ci_t'][0]:.4f}, {info['ci_t'][1]:.4f}] (t={t_crit_reg:.4f})")

print("\nPaired differences (after - before):")
print(f" mean diff = {mean_diff:.4f}, sd_diff = {sd_diff:.4f}, SE = {se_mean_diff:.4f}")
print(f" Normal z CI (95%): [{ci_z_diff[0]:.4f}, {ci_z_diff[1]:.4f}] (z={z_crit:.4f})")
print(f" t CI    (95%, df={df_diff}): [{ci_t_diff[0]:.4f}, {ci_t_diff[1]:.4f}] (t={t_crit_diff:.4f})")

# ---- PLOTS ----

# 1) Scatter with OLS regression line and t-based 95% CI band for mean prediction
fig, ax = plt.subplots(figsize=(7,5))
sns.scatterplot(x=x, y=y, s=80, color='C0', edgecolor='k', ax=ax)
ax.set_xlabel("bp_before")
ax.set_ylabel("bp_after")
ax.set_title("bp_after vs bp_before with OLS fit and 95% CI (t-based)")

# Regression line (predicted mean) and t-based CI for the mean prediction using statsmodels get_prediction
pred_X = sm.add_constant(np.linspace(x.min()-5, x.max()+5, 100))
pred_res = model.get_prediction(pred_X)
pred_summary = pred_res.summary_frame(alpha=alpha)  # includes mean, mean_ci_lower/upper (t-based)
mean_pred = pred_summary['mean'].values
ci_lower = pred_summary['mean_ci_lower'].values
ci_upper = pred_summary['mean_ci_upper'].values

# Plot regression line and CI ribbon
ax.plot(pred_X[:,1], mean_pred, color='C1', label='OLS fit')
ax.fill_between(pred_X[:,1], ci_lower, ci_upper, color='C1', alpha=0.25, label='95% CI (t-based)')

# Also annotate slope and intercept with t-CI
slope_info = reg_ci['bp_before']
intercept_info = reg_ci['const']
annot_text = (f"Intercept={intercept_info['est']:.2f} "
              f"[{intercept_info['ci_t'][0]:.2f}, {intercept_info['ci_t'][1]:.2f}]\n"
              f"Slope={slope_info['est']:.3f} "
              f"[{slope_info['ci_t'][0]:.3f}, {slope_info['ci_t'][1]:.3f}]")
ax.text(0.02, 0.98, annot_text, transform=ax.transAxes, fontsize=9,
        verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
ax.legend()
plt.tight_layout()

# 2) Paired differences plot with error bars for Normal and t CIs
fig2, ax2 = plt.subplots(figsize=(6,3))
ax2.axhline(0, color='gray', linestyle='--', linewidth=0.8)
# plot mean point
ax2.errorbar(0, mean_diff, yerr=[[mean_diff - ci_z_diff[0]], [ci_z_diff[1] - mean_diff]],
             fmt='o', color='C2', label='Normal z CI', capsize=8, elinewidth=2)
ax2.errorbar(0.15, mean_diff, yerr=[[mean_diff - ci_t_diff[0]], [ci_t_diff[1] - mean_diff]],
             fmt='o', color='C3', label='t CI (df=n-1)', capsize=8, elinewidth=2)
# add small jitter of raw diffs for context
ax2.scatter(np.random.normal(0, 0.03, size=n), diff, color='k', alpha=0.7, label='individual diffs')
ax2.set_xlim(-0.6, 0.6)
ax2.set_xticks([0, 0.15])
ax2.set_xticklabels(['Normal CI', 't CI'])
ax2.set_ylabel('after - before')
ax2.set_title('Mean paired difference with 95% CIs')
ax2.legend()
plt.tight_layout()

plt.show()
```
